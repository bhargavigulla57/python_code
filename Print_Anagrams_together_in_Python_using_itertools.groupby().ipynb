```python
from itertools import groupby

a = ['bat', 'nat', 'tan', 'ate', 'eat', 'tea']
a.sort(key=lambda x: ''.join(sorted(x)))
groups = [list(g) for _, g in groupby(a, key=lambda x: ''.join(sorted(x)))]

print(groups)
```
