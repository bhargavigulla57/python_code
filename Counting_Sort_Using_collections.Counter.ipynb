```python
from collections import Counter

def count_sort(arr):
    count = Counter(arr)
    res = []
    for num in sorted(count.keys()):
        res += [num] * count[num]
    return res

arr = [4, 2, 2, 8, 3, 3, 1]
sort_arr = count_sort(arr)
print( sort_arr)
```
