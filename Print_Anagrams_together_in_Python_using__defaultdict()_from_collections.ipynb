```python
from collections import Counter, defaultdict

a = ['bat', 'nat', 'tan', 'ate', 'eat', 'tea']
res = defaultdict(list)

for word in a:
    key = tuple(sorted(Counter(word).items()))
    res[key].append(word)

print(list(res.values()))
```
