```python
def shell_half(a):
    n = len(a)
    g = n // 2
    while g > 0:
        for i in range(g, n):
            t = a[i]
            j = i
            while j >= g and a[j-g] > t:
                a[j] = a[j-g]
                j -= g
            a[j] = t
        g //= 2

arr = [12, 34, 54, 2, 3]
shell_half(arr)
print(arr)
```
