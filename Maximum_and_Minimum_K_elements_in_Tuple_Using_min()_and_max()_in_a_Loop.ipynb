```python
tup = (5, 20, 3, 7, 6, 8)
K = 2

m1, m2 = [], []
for elem in tup:
    if len(m1) < K:
        m1.append(elem)
    else:
        if elem < max(m1):
            m1.remove(max(m1))
            m1.append(elem)

    if len(m2) < K:
        m2.append(elem)
    else:
        if elem > min(m2):
            m2.remove(min(m2))
            m2.append(elem)

m1.sort()
m2.sort(reverse=True)

print( m1)
print( m2)
```
